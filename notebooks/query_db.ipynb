```python
import sqlite3

db_path = "../backend/src/db/recipes.db"

conn = sqlite3.connect(db_path)
cursor = conn.cursor()

tables = ["recipe_meta", "ingredients", "recipe_ingredients", "recipe_steps", "tags", "recipe_tags"]
for table in tables:
    cursor.execute(f"SELECT * FROM {table}")
    print(f"Contents of {table}: {cursor.fetchall()}")
```
