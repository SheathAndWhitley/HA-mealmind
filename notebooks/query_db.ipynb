```python
import sqlite3

db_path = "../backend/src/db/recipes.db"

conn = sqlite3.connect(db_path)
cursor = conn.cursor()

tables = ["recipe_meta", "ingredients", "recipe_ingredients", "recipe_steps", "tags", "recipe_tags"]
for table in tables:
    cursor.execute(f"SELECT * FROM {table}")
    print(f"Contents of {table}:\n {cursor.fetchall()}")
```

```python
# Select three random recipes for the telegram bot, test -> moved to recipe_picker.py
import sqlite3

conn = sqlite3.connect(db_path)
conn.row_factory = sqlite3.Row # each row acts as a dict, not needing index
cursor = conn.cursor()


query = """SELECT recipe_id, title, description, preptime_minutes, cooktime_minutes, continent 
            FROM recipe_meta ORDER BY RANDOM() LIMIT 3 
            """

cursor.execute(query)
recipes = cursor.fetchall()

dict_recipes = []
for row in recipes:
    dict_recipes.append(dict(row))

print(dict_recipes)
```
