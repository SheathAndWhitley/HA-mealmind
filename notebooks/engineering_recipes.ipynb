```python
import sqlite3
import pathlib
import ollama

cwd = pathlib.Path.cwd()

db_path = "/home/kristian/Coding/HA-mealmind/backend/src/db/recipes.db"

def fetch_all_recipes(db_path: pathlib.Path | str) -> list[dict]:
    conn = sqlite3.connect(db_path)

    conn.row_factory = sqlite3.Row
    cursor = conn.cursor()

    cursor.execute("SELECT * FROM recipe_meta")
    recipes = [dict(row) for row in cursor.fetchall()]

    for recipe in recipes:
        recipe_id = recipe["title"]
        print(recipe_id)

        
fetch_all_recipes(db_path)
```
